# Mission 4 : What does the np.empty() method do? Don't just copy the documentation, explain in your own words here

`np.empty` crée un array de la taille demandée **sans initialiser ses valeurs** : il contient ce qui traînait déjà en mémoire (des valeurs quelconques, parfois des 0, mais rien de garanti). C'est plus rapide que `np.zeros` car numpy n'écrit rien dedans. 
c'est utile quand on veut créer un tableau que l'on va remplir ensuite. 

```python
# Mission 5 : Create an array filled with only zeros. You can choose the length of this array.
arr = np.zeros(100)
print(arr)
```

```python
# Mission 6 : Create an array filled with 10 random integers.
arr = np.random.randint(0, 100, size=10)
print(arr)
```

```python
# Mission 7 : Create an array that contains 50 values randomly selected only from these numbers: 0, 1, 2, 3, 4, 5, 6, 7.
arr = np.random.choice([0,1,2,3,4,5,6,7], size=50)
print(arr)
```
