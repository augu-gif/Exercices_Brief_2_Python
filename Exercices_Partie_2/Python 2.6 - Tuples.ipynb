# Mission 4

Create 2 tuples of the same length that contain only integers.

Then create a script that compares the sum of their elements and displays the tuple that has the highest total.

```python
tuple_1 =(15,858,54,65,4,654,2)
tuple_2 =(156,8,87,9,987,987)
somme_a = sum(tuple_1)
somme_b = sum(tuple_2)
if somme_a > somme_b:
    print("le tuple_1 a une somme plus élevé")
elif somme_a < somme_b:
    print("le tuple_2 a une somme plus élevé")
else: 
    print("les deux tuples ont la même somme")
```

# Mission 5

Here's a tuple:

`my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")`

Ask the user for a string, then for an integer position. Create an altered copy of `my_tuple` where the element at the given position is now the given string.

```python
my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")
string = input("tapez votre mot :")
position = int(input("tapez votre position :"))-1
new_tuple_2 = my_tuple[:position] + (string,) + my_tuple[position:]
print(new_tuple_2)
```
