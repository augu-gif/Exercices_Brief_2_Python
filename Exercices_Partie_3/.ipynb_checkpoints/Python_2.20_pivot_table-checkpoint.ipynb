# Execute the code below

```python
import pandas as pd
import numpy as np
link = "https://raw.githubusercontent.com/WildCodeSchool/data-training-resources/main/quests/weather2019.csv"
df_weather = pd.read_csv(link)
```

```python
#columns = colonnes
#index = lignes
#values = valeurs
#aggfunc = fonction d'agrégation
```

# You have now a DataFrame with lots of weather data

```python
df_weather.head()
```

# PIVOT TABLE

## It's like in Excel, but with Pandas

There are 2 syntaxes, choose your preferred :
- As a method : YOUR_DATAFRAME.pivot_table()
- As a function : pd.crosstab()

```python
# Example
df_weather.pivot_table(values = 'MAX_TEMPERATURE_C', index = 'MONTH', aggfunc = 'mean')
```

## Create a Pivot table with :
- Months on rows
- the maximum of Max Temperature on values
- Opinion on columns

```python
# Hop !
df_weather.pivot_table(values = 'MAX_TEMPERATURE_C', index = 'MONTH', columns = "OPINION", aggfunc = "max")
```

## Pivot table with :
- Days on rows
- Precipitation on values
- Months on columns


```python
# Hop !
df_weather.pivot_table(values = 'PRECIP_TOTAL_DAY_MM', index = 'DAY', columns = 'MONTH')
```

## Pivot table with :
- Months on rows
- the sum of Precipitations on values

**Then display it as a column chart.** (optionnal)

```python
# Hop !
precip_par_mois = df_weather.pivot_table(values = 'PRECIP_TOTAL_DAY_MM', index = 'MONTH', aggfunc = "sum")
precip_par_mois.plot(kind = "bar", legend = False, title = "Précipitations totales par mois (mm)")
precip_par_mois
```

## Pivot table with :
- Months on rows
- the average of Max Temperature on a first column of values
- the average of Sunhours on a second column of values

```python
# Hop !
df_MTS = df_weather.pivot_table(index="MONTH",values=["MAX_TEMPERATURE_C", "SUNHOUR"],aggfunc="mean")
df_MTS
```

# Chart
Using your favorite library (MatPlotLib, Seaborn, Plotly or what you want), draw a chart from the previous pivot table with :
- Temperature in a blue column chart
- Sunhours in a red line plot

Tip : Once your pivot table created, you can help yourself with the `.reset_index()`  method.

```python
# Your code here
import matplotlib.pyplot as plt

df_plot = df_MTS.reset_index()

fig, ax1 = plt.subplots(figsize=(10, 5))
# températures : colonnes bleues (axe de gauche)
ax1.bar(df_plot["MONTH"], df_plot["MAX_TEMPERATURE_C"], color="blue")
ax1.set_xlabel("Mois")
ax1.set_ylabel("Température max moyenne (°C)", color="blue")
ax1.set_xticks(df_plot["MONTH"])

# heures de soleil : ligne rouge (axe de droite)
ax2 = ax1.twinx()
ax2.plot(df_plot["MONTH"], df_plot["SUNHOUR"], color="red", marker="o")
ax2.set_ylabel("Heures de soleil moyennes", color="red")

plt.title("Température max et heures de soleil moyennes par mois (2019)")
plt.show()
```
