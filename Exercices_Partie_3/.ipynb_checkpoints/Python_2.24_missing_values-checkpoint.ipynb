# Function to automatic merge and fill missing values



## Create the function
You have to create a function which take 2 arguments :
- argument 1 : DataFrame main
- argument 2 : DataFrame opinion

Your function will return a new DataFrame which is the left merge of both DataFrames, and which has no missing values (so your function has to fill them), add new columns (flag, MONTH and DAY), and OPINION should be translated.

So you have to copy each useful code from the previous quests, and integrate it into a function. **This should achieve the same results as the previous quests.**

```python
import pandas as pd
```

```python
# Chargement des Datasets

link_main = "https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather_main_2018.csv"
link_opinion = "https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather_opinion_2018.csv"
df_main = pd.read_csv(link_main)
df_opinion = pd.read_csv(link_opinion)
```

```python
# Création d'une fonction merge_and_fill qui prend en paramètre
# deux DataFrame pour les joindre avec un left join sur la colonne de date
# (mêmes choix de remplissage que dans la quête précédente 2.18)

def merge_and_fill(DataFrameMain, DataFrameOpinion):
  DataFrameOpinion = DataFrameOpinion.rename(columns={"date": "DATE"})
  # MERGE BOTH DATAFRAMES
  NewDataFrame = pd.merge(DataFrameMain, DataFrameOpinion, on="DATE", how="left")

  # Ajout de la colonne FLAG AVANT de remplir : True = ligne dont on va compléter les valeurs
  NewDataFrame["flag"] = NewDataFrame["OPINION"].isna()

  # FILL IN MISSING VALUES
  # WEATHER_CODE_EVENING => médiane (c'est un code : la médiane donne un code qui existe)
  median = NewDataFrame["WEATHER_CODE_EVENING"].median()
  NewDataFrame["WEATHER_CODE_EVENING"] = NewDataFrame["WEATHER_CODE_EVENING"].fillna(median)
  # TOTAL_SNOW_MM, UV_INDEX et SUNHOUR => propagation de la valeur précédente (ffill)
  # puis bfill pour le cas où c'est le 1er janvier qui manque (rien avant lui à propager, ex. en 2013)
  colonnes_propagees = ["TOTAL_SNOW_MM", "UV_INDEX", "SUNHOUR"]
  NewDataFrame[colonnes_propagees] = NewDataFrame[colonnes_propagees].ffill().bfill()
  # OPINION => "Unknown" pour les valeurs manquantes
  NewDataFrame["OPINION"] = NewDataFrame["OPINION"].fillna("Unknown")

  # Conversion de la colonne DATE au format Datetime
  NewDataFrame["DATE"] = pd.to_datetime(NewDataFrame["DATE"])
  # Création des colonnes MONTH et DAY
  NewDataFrame["MONTH"] = NewDataFrame["DATE"].dt.month
  NewDataFrame["DAY"] = NewDataFrame["DATE"].dt.day

  # Traduction de la colonne OPINION en français
  traduction = {
      "very bad": "très mauvais",
      "bad": "mauvais",
      "neutral": "neutre",
      "not good not bad": "neutre",
      "good": "bon",
      "very good": "très bon",
      "Unknown": "inconnue",
  }
  NewDataFrame["OPINION"] = NewDataFrame["OPINION"].replace(traduction)

  # Return new merged dataframe
  return NewDataFrame
```

```python
df_2018 = merge_and_fill(df_main, df_opinion)
df_2018
```

## Execution with a loop
You will execute your function in a 9 rounds loop, once per year, and agregate this in a new big DataFrame with 24 columns and lots of rows.

```python
# Chaque année entre 2010 et 2018 (inclus) a deux 2 sources : weather_main_YYYY.csv and weather_opinion_YYYY.csv
# Charger les deux datasets et joignez-les grâce à la fonction merge_and_fill
# Concaténez ce nouveau DataFrame au(x) précédent(s) pour n'obtenir qu'un long DataFrame df_weather

def load_weather():
    dfs_annees = []
    for annee in range(2010, 2019):  # 2010 à 2018 inclus
        # 1. Charger le fichier weather_main de l'année
        df_main = pd.read_csv(f"https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather_main_{annee}.csv")
        # 2. Charger le fichier weather_opinion de l'année
        df_opinion = pd.read_csv(f"https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather_opinion_{annee}.csv")
        # 3. Fusionner les deux DataFrames
        dfs_annees.append(merge_and_fill(df_main, df_opinion))

    # 4. Concaténer tous les DataFrames
    return pd.concat(dfs_annees, ignore_index=True)
```

```python
# Affichage des premières lignes de df_weather (2010 à 2018)
df_weather_2010_2018 = load_weather()
print(df_weather_2010_2018.shape)
df_weather_2010_2018.head()
```

```python
# Création d'une fonction clean qui :
# * crée une colonne FLAG indiquant une valeur manquante
# * convertit la colonne DATE au format datetime
# * traduit l'opinion en français
def clean(DataFrameYear):

  # ALIGN DATAFRAME SHAPE, DATATYPE
  # Colonne FLAG
  DataFrameYear['flag'] = DataFrameYear.isna().any(axis=1)

  # Colonne DATE
  DataFrameYear["DATE"] = pd.to_datetime(DataFrameYear["DATE"])
  # Colonnes MONTH et DAY (absentes des fichiers 2021 et 2022) recalculées depuis la date
  DataFrameYear["MONTH"] = DataFrameYear["DATE"].dt.month
  DataFrameYear["DAY"] = DataFrameYear["DATE"].dt.day
  # Colonne OPINION
  traduction = {
      "météo très défavorable": "très mauvais",
      "météo défavorable": "mauvais",
      "météo correcte": "neutre",
      "météo favorable": "bon",
      "météo idéale": "très bon",
      "very bad": "très mauvais",
      "bad": "mauvais",
      "neutral": "neutre",
      "not good not bad": "neutre",
      "good": "bon",
      "very good": "très bon",
      "Unknown": "inconnue",
  }
  DataFrameYear["OPINION"] = DataFrameYear["OPINION"].replace(traduction)
  # Return new cleaned dataframe
  return DataFrameYear
```

```python
# On crée une liste de DataFrames pour les années 2019 à 2022 (inclus)

links = {}
for year in range(2019, 2023):
    # Collect both DataFrames each year
    year = str(year)
    links[year] = "https://raw.githubusercontent.com/WildCodeSchool/data-training-resources/main/quests/weather" + year + ".csv"

# Create dataframes from the links to clean
dfs = {}  # Dictionary to store dataframes
for year, link in links.items():
    dfs[year] = pd.read_csv(link)
    dfs[year] = clean(dfs[year])

dfs
```

```python
# Concaténez df_weather avec df_2019 à 2022
# (on repart toujours de df_weather_2010_2018 : la cellule peut être relancée sans dupliquer les lignes)
df_weather_2019_2022 = pd.concat(dfs.values(), ignore_index=True)
# ALIGN DATAFRAME SHAPE : on garde uniquement les colonnes de 2010-2018
# (les fichiers récents ont des colonnes en plus : SUNRISE, SUNSET, WEATHER_CODE_MORNING...)
df_weather_2019_2022 = df_weather_2019_2022[df_weather_2010_2018.columns]
df_weather = pd.concat([df_weather_2010_2018, df_weather_2019_2022], ignore_index=True)
df_weather = df_weather.sort_values("DATE", ignore_index=True)
print(df_weather.shape)
print("valeurs manquantes :", df_weather.isna().sum().sum())
```

# Draw a scatterplot
With Date on X-Axis and Max Temperature on Y-axis

The result must have 11 seasonality for the 11 summers/winters alternance.

Utilisez Seaborn ou Plotly Express selon votre préférence




```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df_weather, x="DATE", y="MAX_TEMPERATURE_C")
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.ylabel("Température maximale (°C)")
plt.title("Évolution des températures maximales")
plt.tight_layout()
plt.show()
```


Find appropriate dataviz to answer to this question :

Which year had the best weather? Are years very different or not ?


```python
df_weather["YEAR"] = df_weather["DATE"].dt.year
sns.boxplot(data=df_weather, x="YEAR", y="MAX_TEMPERATURE_C")
plt.title("Distribution des températures maximales par année")
plt.xlabel("Année")
plt.ylabel("Température maximale (°C)")
plt.show()
```

Quelles années semblent avoir bénéficié des meilleures conditions météo ? Pour quelles raisons ?

```python
# Répartition des opinions (en %) par année
opinions_par_annee = pd.crosstab(df_weather["YEAR"], df_weather["OPINION"], normalize="index") * 100
display(opinions_par_annee.round(1))

# Part des jours jugés "bon" ou "très bon" par année
df_weather["OPINION_POSITIVE"] = df_weather["OPINION"].isin(["bon", "très bon"])
(df_weather.groupby("YEAR")["OPINION_POSITIVE"].mean() * 100).sort_values(ascending=False)
```

En proportion de jours jugés bon ou très bon , 2022 (≈ 37 %) et 2014 (≈ 35 %) ressortent comme les meilleures années, et 2016 (≈ 19 %) comme la moins bonne.Le boxplot montre aussi des distributions de températures maximales très similaires d'une année à l'autre.
